# Ramsey Interferometer Simulation

This notebook simulates Ramsey interferometry for a two-level quantum system.

```python
import numpy as np
import matplotlib.pyplot as plt

# Detuning range
delta = np.linspace(-10, 10, 1000)
T = 1.0  # free evolution time

# Ramsey fringe probability
P = 0.5 * (1 + np.cos(delta * T))

plt.figure(figsize=(8,4))
plt.plot(delta, P)
plt.xlabel('Detuning')
plt.ylabel('Excited-state Probability')
plt.title('Ramsey Interference Fringes')
plt.grid(True)
plt.show()
```
